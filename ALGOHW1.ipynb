# Problem 2: On Sorting

## 1. Algorithm

$$
m=\left\lceil\frac{2n}{3}\right\rceil.
$$

For a list of size at most one, do nothing. For a list of size
two, compare the two elements and swap them if necessary.
For a larger list, recursively sort the first $m$ elements,
then the last $m$ elements, and finally the first $m$ elements
again. The algorithm operates in place, using inclusive indices
to specify each subarray without copying it.

## 2. Pseudocode

```text
THREE-PHASE-SORT(A, left, right)
    n = right - left + 1

    if n <= 1
        return

    if n == 2
        if A[left] > A[right]
            swap A[left] and A[right]
        return

    m = ceil(2 * n / 3)

    THREE-PHASE-SORT(A, left, left + m - 1)
    THREE-PHASE-SORT(A, right - m + 1, right)
    THREE-PHASE-SORT(A, left, left + m - 1)
```

For a list with indices $1,\ldots,n$, the initial call is
`THREE-PHASE-SORT(A, 1, n)`.

## 3. Correctness

$$
k=\left\lfloor\frac n3\right\rfloor,
\qquad
m=\left\lceil\frac{2n}{3}\right\rceil=n-k.
$$

$$
A=L\,\Vert\,M\,\Vert\,R,
\qquad
|L|=|R|=k,
\qquad
|M|=n-2k\ge k.
$$

$$
L\Vert M=A[1:m],
\qquad
M\Vert R=A[k+1:n].
$$

Here, $\Vert$ denotes concatenation, and indices in this proof
are relative to the current subarray.

We prove correctness by strong induction on the subarray size $n$.

**Base cases.**
A subarray of size one is already sorted.
For size two, the algorithm swaps the elements exactly when
they are out of order, so the result is sorted.

**Inductive hypothesis.**
Assume that the algorithm correctly sorts every subarray
of size less than $n$.

**Inductive step.**
Consider a subarray of size $n \ge 3$.
Since $m<n$, all three recursive calls correctly sort their
respective subarrays by the inductive hypothesis.

After the first call, $L\Vert M$ is sorted. Since the elements
are distinct, every element of $L$ is smaller than every
element of $M$. Because $M$ contains at least $k$ elements,
each element of $L$ has at least $k$ elements larger than it.
Thus, none of the largest $k$ elements of the entire subarray
can be in $L$. All of them must be in $M\Vert R$.

The second call sorts $M\Vert R$. Therefore, it places the
largest $k$ elements of the entire subarray in $R$, in
increasing order.

The third call sorts $L\Vert M$ without changing $R$.
Every element of $L\Vert M$ is smaller than every element
of $R$, and both parts are now sorted. Hence, the entire
subarray is sorted.

The algorithm preserves all input elements because its only
operation that changes the array is a swap. Also, recursive
subarrays are strictly smaller whenever $n\ge 3$, so the
algorithm terminates.

Therefore, the algorithm correctly sorts the input list.

## 4. Running Time

$$
T(n)=
\begin{cases}
\Theta(1), & n\le 2,\\[3pt]
3T\!\left(\left\lceil\frac{2n}{3}\right\rceil\right)
+\Theta(1), & n\ge 3.
\end{cases}
$$

$$
a=3,
\qquad
b=\frac32,
\qquad
f(n)=\Theta(1).
$$

$$
p=\log_b a=\log_{3/2}3\approx 2.7095.
$$

$$
f(n)=\Theta(1)=O(n^{p-\varepsilon}),
\qquad
\varepsilon=\frac p2>0.
$$

$$
T(n)=\Theta\!\left(n^{\log_{3/2}3}\right)
$$

Each non-base call makes three recursive calls on subarrays
of size $\lceil 2n/3\rceil$. Apart from those calls, it performs
only constant work: checking the size, computing indices,
and returning. No subarrays are copied.

Applying Case 1 of the Master Theorem gives the stated tight
bound. The ceiling in the subproblem size does not change
the asymptotic result.

## 5. Practical Use

$$
\log_{3/2}3>2,
\qquad
n\log n=o\!\left(n^{\log_{3/2}3}\right).
$$

I would not use this algorithm in practice.
Its running time grows faster than quadratic time, whereas
merge sort and heapsort have worst-case running time
$\Theta(n\log n)$. Those algorithms are asymptotically
more efficient.

# Problem 3: On Measuring Disorder

$$
\operatorname{Inv}(A)
=
\left|\left\{(i,j):1\le i<j\le n,\ A[i]>A[j]\right\}\right|.
$$

## Part (a): Brute-Force Algorithm

### Algorithm Description

Examine every pair of indices with the first index smaller
than the second. Increment a counter whenever the first
element is larger than the second, and return the counter.

### Pseudocode

```text
BRUTE-FORCE-COUNT(A)
    n = length(A)
    count = 0

    for i = 1 to n - 1
        for j = i + 1 to n
            if A[i] > A[j]
                count = count + 1

    return count
```

### Correctness

$$
\text{count}
=
\sum_{i=1}^{n-1}
\sum_{j=i+1}^{n}
\mathbf{1}\{A[i]>A[j]\}
=
\operatorname{Inv}(A).
$$

The nested loops examine every pair of indices satisfying
$i<j$ exactly once. The counter increases exactly when the
examined pair satisfies $A[i]>A[j]$, which is precisely the
definition of an inversion. Therefore, every inversion is
counted once, and no other pair is counted. The algorithm
returns the correct number of inversions.

### Complexity

$$
\sum_{i=1}^{n-1}(n-i)
=
\frac{n(n-1)}{2}.
$$

$$
\boxed{T(n)=\Theta(n^2)},
\qquad
\boxed{S(n)=\Theta(1)}.
$$

The algorithm examines exactly $n(n-1)/2$ pairs and performs
constant work for each pair. It uses only a counter and
loop indices, so its auxiliary space is constant.

## Part (b): An O(n log n) Algorithm

### Algorithm Description

Split the array into two contiguous halves. Recursively
sort each half and count its internal inversions.

Then merge the two sorted halves while counting cross
inversions, which have one element from each half.
If the next left element is larger than the next right
element, every remaining left element forms an inversion
with that right element. Add the number of remaining left
elements to the count.

Return the merged sorted array together with the sum of
the left, right, and cross inversion counts.

### Pseudocode

All arrays use 1-based indexing. Appending an element takes
constant time, using preallocated output space.

```text
SORT-AND-COUNT(A)
    n = length(A)

    if n <= 1
        return (A, 0)

    mid = floor(n / 2)

    (L, leftCount) = SORT-AND-COUNT(A[1 .. mid])
    (R, rightCount) = SORT-AND-COUNT(A[mid + 1 .. n])

    (B, crossCount) = MERGE-AND-COUNT(L, R)

    total = leftCount + rightCount + crossCount
    return (B, total)
```

```text
MERGE-AND-COUNT(L, R)
    p = length(L)
    q = length(R)
    i = 1
    j = 1
    crossCount = 0
    B = empty output array with capacity p + q

    while i <= p and j <= q
        if L[i] <= R[j]
            append L[i] to B
            i = i + 1
        else
            append R[j] to B
            crossCount = crossCount + (p - i + 1)
            j = j + 1

    while i <= p
        append L[i] to B
        i = i + 1

    while j <= q
        append R[j] to B
        j = j + 1

    return (B, crossCount)
```

Call `SORT-AND-COUNT(A)` and return the count in its result.

### Correctness

$$
I(A)=I_{\mathrm{left}}+I_{\mathrm{right}}+I_{\mathrm{cross}}.
$$

$$
L[i]>R[j]
\quad\Longrightarrow\quad
L[t]>R[j]
\quad\text{for every }t\in\{i,\ldots,p\}.
$$

$$
\Delta I_{\mathrm{cross}}=p-i+1.
$$

**Correctness of the merge step.**

Every element of the original left half precedes every
element of the original right half. Thus, a pair containing
one element from each half is an inversion exactly when
the left element is larger than the right element.

The merge procedure repeatedly appends the smaller of the
two current elements, so it produces a sorted array
containing exactly the elements of both input arrays.

If $L[i]\le R[j]$, then $L[i]$ is no larger than any remaining
right element, because $R$ is sorted. Hence, taking $L[i]$
creates no uncounted cross inversion.

If $L[i]>R[j]$, all remaining left elements are larger than
$R[j]$, because $L$ is sorted. There are exactly $p-i+1$
such elements. Previously appended left elements are
smaller than $R[j]$, so the increment counts exactly the
cross inversions involving this right element.

Each right element is processed once, so no cross inversion
is counted twice. If the left array is exhausted, no left
elements remain to form additional inversions. If the right
array is exhausted, all cross inversions have already been
counted. Therefore, the merge procedure returns the exact
cross inversion count.

**Correctness of the recursive algorithm.**

We prove by strong induction on $n$ that `SORT-AND-COUNT`
returns both a sorted version of its input and its exact
inversion count.

For $n\le 1$, the input is already sorted and contains no
inversions, so the returned result is correct.

For $n\ge 2$, assume the claim holds for all smaller arrays.
Both recursive calls operate on smaller arrays, so they
return correctly sorted halves and correct internal
inversion counts.

By the merge argument, the merge step returns a sorted
version of the entire array and the correct cross inversion
count. Every original inversion lies entirely in the left
half, entirely in the right half, or crosses between them.
These categories are disjoint and exhaustive. Their sum
therefore equals the total inversion count.

Thus, the algorithm is correct.

### Running Time

$$
T(n)=
\begin{cases}
\Theta(1), & n\le 1,\\[3pt]
T(\lfloor n/2\rfloor)+T(\lceil n/2\rceil)+\Theta(n),
& n\ge 2.
\end{cases}
$$

For powers of two, this becomes

$$
T(n)=2T(n/2)+\Theta(n).
$$

$$
a=2,\qquad b=2,\qquad n^{\log_b a}=n.
$$

$$
\boxed{T(n)=\Theta(n\log n)}.
$$

The algorithm makes two recursive calls on approximately
equal halves. Merging processes each element once, so the
merge takes linear time. Creating the two halves, if copied,
also takes only linear time.

By Case 2 of the Master Theorem, the running time is
$\Theta(n\log n)$. Floors and ceilings do not change this
asymptotic bound.

### Auxiliary Space

$$
S_{\mathrm{arrays}}(n)=O(n),
\qquad
S_{\mathrm{stack}}(n)=O(\log n).
$$

$$
\boxed{S(n)=O(n)}.
$$

The merge arrays require linear auxiliary space, and the
recursion has logarithmic depth. With sequential recursive
calls and temporary arrays released when no longer needed,
the peak total auxiliary space is linear.
