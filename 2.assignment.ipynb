## 1. Chain Rule & Weight Updates

The network in the lecture is defined as:

$$
z = w_1x+b_1,\qquad h=\operatorname{ReLU}(z)
$$

$$
y=w_2h+b_2
$$

The loss function is:

$$
L=\frac{1}{2}(y-t)^2
$$

$y$ is the prediction and $t$ is the target value.

To find the gradient with respect to $w_1$, I first look at how $w_1$ affects the loss. Changing $w_1$ changes $z$, which affects $h$, then $y$, and finally $L$. So the chain rule gives:

$$
\frac{\partial L}{\partial w_1}
=
\frac{\partial L}{\partial y}·
\frac{\partial y}{\partial h}·
\frac{\partial h}{\partial z}·
\frac{\partial z}{\partial w_1}
$$


The individual derivatives are:

$$
\frac{\partial L}{\partial y}
=
\frac{1}{2}\cdot 2(y-t)
=
y-t
$$

$$
\frac{\partial y}{\partial h}=w_2
$$

For ReLU, the output equals $z$ when $z$ is positive, and it is
zero when $z$ is negative. Therefore:

$$
\frac{\partial h}{\partial z}
=
\begin{cases}
1, & z>0,\\
0, & z<0.
\end{cases}
$$

At $z=0$, ReLU is not differentiable. I use the usual convention of setting its derivative to zero there for the weight update.

Finally, since $z=w_1x+b_1$:

$$
\frac{\partial z}{\partial w_1}=x
$$

Multiplying these terms gives:

$$
\frac{\partial L}{\partial w_1}
=
(y-t)w_2\operatorname{ReLU}'(z)x
$$


With the convention at zero, this can also be written as:

$$
\frac{\partial L}{\partial w_1}
=
\begin{cases}
(y-t)w_2x, & w_1x+b_1>0,\\
0, & w_1x+b_1\leq0.
\end{cases}
$$

One thing I notice is that if $z<0$, the gradient is zero even if the prediction is wrong. This is because the ReLU derivative blocks the gradient through this hidden node.


## 2. Gradient Descent

Gradient descent uses the gradient to decide how to change the weights. The gradient points toward increasing loss, so we move in the opposite direction. With a small enough step, this can reduce the loss.

The update rule is:

$$
w_{\text{new}}
=
w_{\text{old}}
-
\eta\frac{\partial L}{\partial w}
$$

The learning rate $\eta$ controls how big the update is. The gradient is calculated using the current weights.


For $w_1$ in this problem, the rule becomes:

$$
w_{1,\text{new}}
=
w_{1,\text{old}}
-
\eta(y-t)w_2\operatorname{ReLU}'(z)x
$$

If the gradient is positive, we decrease the weight. If it is
negative, we increase the weight.

### Explain why the learning rate is too large：

For the same gradient, a larger learning rate produces a larger step. If the step is too large, the weight may move past the minimum to the other side. The gradient may then reverse direction, causing the next update to cross the minimum again.

Repeated overshooting can cause oscillation. If the oscillations do not shrink, gradient descent fails to converge. If the updates move the weight progressively farther from the minimum, the process diverges.


For example, take a simple loss:

$$
L(w)=\frac{1}{2}w^2,\qquad \frac{dL}{dw}=w
$$

Then:

$$
w_{\text{new}}=(1-\eta)w_{\text{old}}
$$

If $\eta=2.5$ and we start at $w=1$, the updates are:

$$
1\rightarrow -1.5\rightarrow 2.25\rightarrow -3.375
$$

The minimum is at zero, but each update moves farther away from it. So in this example, gradient descent diverges. Crossing the minimum does not always mean the method will fail. It can still converge if the back-and-forth movements get smaller. The issue with a learning rate that is too large is that these movements may stay the same size or keep growing.
